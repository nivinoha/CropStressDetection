```python
import os
import requests
import pandas as pd

# ✅ Step 1: Prepare folder path
save_dir = r"C:\Users\HP\CropStressDetection\data\weather"
os.makedirs(save_dir, exist_ok=True)

# ✅ Step 2: Define location & time range
lat, lon = 30.9, 75.85   # Punjab region
start = "20240101"
end   = "20241231"

# ✅ Step 3: NASA POWER API endpoint (correct)
url = (
    f"https://power.larc.nasa.gov/api/temporal/daily/point"
    f"?parameters=T2M,RH2M,PRECTOT"
    f"&community=AG"
    f"&start={start}"
    f"&end={end}"
    f"&latitude={lat}"
    f"&longitude={lon}"
    f"&format=JSON"
)

print("Fetching data...")
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    print("✅ Data received successfully")

    # ✅ Extract each parameter dictionary
    params = data["properties"]["parameter"]
    df = pd.DataFrame({
        "Date": list(params["T2M"].keys()),
        "T2M": list(params["T2M"].values()),
        "RH2M": list(params["RH2M"].values()),
        "PRECTOT": list(params["PRECTOT"].values()),
    })

    # ✅ Convert Date & add stress index
    df["Date"] = pd.to_datetime(df["Date"])
    df["stress_index"] = df["T2M"] * (1 - df["RH2M"] / 100)

    # ✅ Save CSV
    save_path = os.path.join(save_dir, "weather_data.csv")
    df.to_csv(save_path, index=False)

    print(f"\n✅ Data saved successfully at: {save_path}")
    display(df.head())
else:
    print("❌ Failed to fetch data.")
    print("Status Code:", response.status_code)
    print("Response:", response.text[:300])
```

```python
import os
import requests
import pandas as pd

# ✅ Step 1: Prepare folder path
save_dir = r"C:\Users\HP\CropStressDetection\data\weather"
os.makedirs(save_dir, exist_ok=True)

# ✅ Step 2: Define location & time range
lat, lon = 30.9, 75.85   # Punjab region
start = "20240101"
end   = "20241231"

# ✅ Step 3: NASA POWER API endpoint
url = (
    f"https://power.larc.nasa.gov/api/temporal/daily/point"
    f"?parameters=T2M,RH2M,PRECTOT"
    f"&community=AG"
    f"&start={start}"
    f"&end={end}"
    f"&latitude={lat}"
    f"&longitude={lon}"
    f"&format=JSON"
)

print("Fetching data...")
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    print("✅ Data received successfully")

    # ✅ Extract available parameters safely
    params = data["properties"]["parameter"]
    available = list(params.keys())
    print(f"Available parameters: {available}")

    # Start with Date column from first available parameter
    first_param = available[0]
    df = pd.DataFrame({"Date": list(params[first_param].keys())})

    # Add all available parameters dynamically
    for key in ["T2M", "RH2M", "PRECTOT"]:
        if key in params:
            df[key] = list(params[key].values())
        else:
            df[key] = None  # fill missing ones with None

    # ✅ Convert Date & calculate stress index
    df["Date"] = pd.to_datetime(df["Date"])
    df["stress_index"] = df["T2M"].astype(float) * (1 - df["RH2M"].astype(float) / 100)

    # ✅ Save CSV
    save_path = os.path.join(save_dir, "weather_data.csv")
    df.to_csv(save_path, index=False)

    print(f"\n✅ Data saved successfully at: {save_path}")
    display(df.head())
else:
    print("❌ Failed to fetch data.")
    print("Status Code:", response.status_code)
    print("Response:", response.text[:300])

    
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load weather data
df = pd.read_csv("../data/weather/weather_data.csv")

# Fill missing values
df.fillna(method='ffill', inplace=True)

# Normalize numeric columns
scaler = MinMaxScaler()
df[['T2M', 'RH2M', 'stress_index']] = scaler.fit_transform(df[['T2M', 'RH2M', 'stress_index']])

df.to_csv("../data/weather/weather_data_clean.csv", index=False)
print("✅ Cleaned and normalized weather data saved.")
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

data_dir = "../data/satellite"

datagen = ImageDataGenerator(
    rescale=1./255,            # normalize pixels
    rotation_range=20,         # random rotation
    zoom_range=0.2,            # random zoom
    horizontal_flip=True,      # flip images
    validation_split=0.2       # split 80/20
)

train_data = datagen.flow_from_directory(
    data_dir,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='training'
)

val_data = datagen.flow_from_directory(
    data_dir,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)
```
